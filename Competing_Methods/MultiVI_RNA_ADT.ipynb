```python
import pandas as pd
import numpy as np
import scvi
from mudata import MuData
import scanpy as sc
import scib
```

```python
adata_f = sc.read('./Data/RNA_ADT/neurIPS/GSE194122_openproblems_neurips2021_cite_BMMC_processed.h5ad')
adata_gex = adata_f
adata_adt = sc.AnnData(X = adata_f.obsm['protein_counts'])
```

```python
adata_rna = adata_gex
adata_protein = adata_adt
mdata = MuData({"rna": adata_rna, "protein": adata_protein})
scvi.model.MULTIVI.setup_mudata(mdata, batch_key="batch",
modalities={"rna_layer": "rna", "protein_layer": "protein", "batch_key": "rna"})
vae = scvi.model.MULTIVI(mdata)
vae.train()
```

```python
emb = vae.get_latent_representation()
adata_gex.obsm["latent"] = emb
```

```python
np.save("MultiVI_neurIPS.npy", adata_gex.obsm["latent"])
```
