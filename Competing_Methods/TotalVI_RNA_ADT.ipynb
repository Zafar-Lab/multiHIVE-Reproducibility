```python
import numpy as np
import scvi
import scanpy as sc
```

```python
adata_f = sc.read('./Data/RNA_ADT/neurIPS/GSE194122_openproblems_neurips2021_cite_BMMC_processed.h5ad')
adata_gex = adata_f
```

```python
sc.pp.normalize_total(adata_gex, target_sum=1e4)
sc.pp.log1p(adata_gex)

sc.pp.highly_variable_genes(
    adata_gex,
    n_top_genes=4000,
    flavor="seurat_v3",
    batch_key="batch",
    subset=True,
    layer="counts",
)

scvi.model.TOTALVI.setup_anndata(
    adata_gex,
    layer="counts",
    batch_key="batch",
    protein_expression_obsm_key="protein_counts"
)
```

```python
vae = scvi.model.TOTALVI(adata_gex,n_latent=60)
vae.train()
```

```python
adata_gex.obsm["latent"] = vae.get_latent_representation()
```

```python
np.save("TotalVI_neurIPS.npy", adata_gex.obsm["latent"])
```
