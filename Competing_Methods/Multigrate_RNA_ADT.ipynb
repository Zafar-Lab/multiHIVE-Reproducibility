```python
import numpy as np
import scanpy as sc
import multigrate as mtg
import muon
```

```python
adata_f = sc.read('./Data/RNA_ADT/neurIPS/GSE194122_openproblems_neurips2021_cite_BMMC_processed.h5ad')
adata_gex = adata_f
adata_adt = sc.AnnData(X = adata_f.obsm['protein_counts'])
```

```python
adata = mtg.data.organize_multiome_anndatas(
    adatas = [[adata_gex], [adata_adt]],            # a list of anndata objects per modality, RNA-seq always goes first
    layers = [['counts'], ['clr']],     # if need to use data from .layers, if None use .X
)
mtg.model.MultiVAE.setup_anndata(
    adata,
    categorical_covariate_keys=["batch"],
    rna_indices_end = 4000,
)
model = mtg.model.MultiVAE(
    adata,
    losses=['nb', 'mse'],
   # z_dim=60,
)
model.train()

model.plot_losses()
model.get_latent_representation()
```
